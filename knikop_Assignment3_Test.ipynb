```python
# Import Libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import sqlite3
```

```python
# Create Database connection
conn = sqlite3.connect("amazon_sales_test.db")
cursor = conn.cursor()
```

```python
# Load the dataset
df = pd.read_csv('amazon.csv')
```

```python
# Clean dataset

# Remove ₹ and commas
df["discounted_price"] = (
    df["discounted_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["actual_price"] = (
    df["actual_price"]
    .str.replace("₹", "", regex=False)
    .str.replace(",", "", regex=False)
    .astype(float)
)

df["discount_percentage"] = (
    df["discount_percentage"]
    .str.replace("%", "", regex=False)
    .astype(float)
)
```

```python
# Verify datatypes
print(df[["discounted_price",
          "actual_price",
          "discount_percentage"
]].dtypes)
```

```python
# Import CSV into SQLite table
df.to_sql("amazon_sales", conn, if_exists="replace", index=False)
```

## Task 2 - Descriptive Statistics(SQL Version)

```python
# Count, Minimum, Maximum, and Average
# For discounted_price, actual_price and discount_percentage

query = """
SELECT
    COUNT(discounted_price) AS count,
    MIN(discounted_price) AS minimum,
    MAX(discounted_price) AS maximum,
    AVG(discounted_price) AS average
FROM amazon_sales;
"""

query1 = """
SELECT
    COUNT(actual_price) AS count,
    MIN(actual_price) AS minimum,
    MAX(actual_price) AS maximum,
    AVG(actual_price) AS average
FROM amazon_sales;
"""

query2 = """
SELECT
    COUNT(discount_percentage) AS count,
    MIN(discount_percentage) AS minimum,
    MAX(discount_percentage) AS maximum,
    AVG(discount_percentage) AS average
FROM amazon_sales;
"""

display(pd.read_sql(query, conn))
display(pd.read_sql(query1, conn))
display(pd.read_sql(query2, conn))
```

```python
# Median
# For discounted_price, actual_price and discount_percentage

query = """
SELECT discounted_price AS median
FROM amazon_sales
ORDER BY discounted_price
LIMIT 1
OFFSET (
    SELECT COUNT(*) / 2
    FROM amazon_sales
);
"""

query1 = """
SELECT actual_price AS median
FROM amazon_sales
ORDER BY actual_price
LIMIT 1
OFFSET (
    SELECT COUNT(*) / 2
    FROM amazon_sales
);
"""

query2 = """
SELECT discount_percentage AS median
FROM amazon_sales
ORDER BY discount_percentage
LIMIT 1
OFFSET (
    SELECT COUNT(*) / 2
    FROM amazon_sales
);
"""

display(pd.read_sql(query, conn))
display(pd.read_sql(query1, conn))
display(pd.read_sql(query2, conn))
```

```python
# Variance
# For discounted_price, actual_price and discount_percentage

query = """
SELECT
AVG(
    (discounted_price - (
        SELECT AVG(discounted_price)
        FROM amazon_sales
    )) *
    (discounted_price - (
        SELECT AVG(discounted_price)
        FROM amazon_sales
    ))
) AS variance
FROM amazon_sales;
"""

query1 = """
SELECT
AVG(
    (actual_price - (
        SELECT AVG(actual_price)
        FROM amazon_sales
    )) *
    (actual_price - (
        SELECT AVG(actual_price)
        FROM amazon_sales
    ))
) AS variance
FROM amazon_sales;
"""

query2 = """
SELECT
AVG(
    (discount_percentage - (
        SELECT AVG(discount_percentage)
        FROM amazon_sales
    )) *
    (discount_percentage - (
        SELECT AVG(discount_percentage)
        FROM amazon_sales
    ))
) AS variance
FROM amazon_sales;
"""

display(pd.read_sql(query, conn))
display(pd.read_sql(query1, conn))
display(pd.read_sql(query2, conn))
```

```python
# Standard Deviation
# For discounted_price, actual_price and discount_percentage

# Single query to get variance directly from SQLite
query = """
SELECT
    AVG((discounted_price - (SELECT AVG(discounted_price) FROM amazon_sales)) * 
        (discounted_price - (SELECT AVG(discounted_price) FROM amazon_sales))) AS var_discounted,
    
    AVG((actual_price - (SELECT AVG(actual_price) FROM amazon_sales)) * 
        (actual_price - (SELECT AVG(actual_price) FROM amazon_sales))) AS var_actual,
    
    AVG((discount_percentage - (SELECT AVG(discount_percentage) FROM amazon_sales)) * 
        (discount_percentage - (SELECT AVG(discount_percentage) FROM amazon_sales))) AS var_percentage
FROM amazon_sales;
"""

# Read the single row of variances
df_var = pd.read_sql(query, conn)

# Compute square root using numpy in Python to get standard deviations
df_std = np.sqrt(df_var)
df_std.columns = ['std_discounted_price', 'std_actual_price', 'std_discount_percentage']
display(df_std)
```

## Task 4 - Filtering and Conditional Querries

```python
# 1. Return rows where a numerical column is above its overall average 
query = """
SELECT *
FROM amazon_sales
WHERE rating > (
    SELECT AVG(rating)
    FROM amazon_sales
);
"""

pd.read_sql(query, conn)
```

```python
# 2. Compute the average of a numerical column for only one chosen category 
query = """
SELECT
    AVG(rating) AS average_rating
FROM amazon_sales
WHERE category = 'Computers&Accessories|Accessories&Peripherals|Cables&Accessories|Cables|USBCables';
"""

pd.read_sql(query, conn)
```

```python
# 3. Return the top 10 highest values of a numerical column 
query = """
SELECT *
FROM amazon_sales
ORDER BY actual_price DESC
LIMIT 10;
"""

pd.read_sql(query, conn)
```
